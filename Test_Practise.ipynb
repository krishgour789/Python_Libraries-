```python
import numpy as np
# Perform the following:

# Replace 30 with 99.
# Add 5 to every element.
# Find elements greater than 25.
# Compute the average of the array.
arr = np.array([10, 20, 30, 40, 50])
arr[2]=99
print(arr)
for i in range(len(arr)):
    arr[i]+=5
print(arr)
x = np.where(arr>25)
print(arr[x])
x = np.average(arr)
print(x)
```
