```python
# The copy SHOULD NOT be affected by the changes made to the original array.
arr = np.array([1,2,3,4,5])
x = arr.copy()
arr[0]=34
print(arr)
print(x)
```

```python
# The view SHOULD be affected by the changes made to the original array.arr = np.array([1,2,3,4,5])
x = arr.view()
arr[0]=34
print(arr)
print(x)
```
