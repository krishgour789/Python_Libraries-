```python
# List Comprehensions
# -> List Comprehension conosists of brackets containing an expression followed by a for clause, and zero
# or more for or if clauses.
import random
a = [random.randint(10,100) for n in range(20)]
print(a)
```

```python
a = [random.randint(10,100) for i in range(20)]
print(a)
```

```python
a = [(x,x**2,x**3) for x in range(10)]
print(a)
```

```python
a = [(x**2) for x in range(10)]
print(a)
```

```python
a = [int(i) for i in ['10','20','20']]
print(a)
```

```python
a  = [(i*i) for i in range(1,10)]
print(a)
```

```python
a = [n for n in range(10,100) if n%2==0]
print(a)
```

```python
a = [n for n in range(10,100) if n%2==1]
print(a)
```

```python
a = [10,40,60,20,50,100,30,12,24]
p = [num for num in a if num>20 or num>50]
print(p)
```

```python
a = ['!' if alphabet in 'aeiou' else alphabet for alphabet in 'Technical']
print(a)
a = ['!' if alphabet in 'aeiou' else alphabet for alphabet in 'KrishGour']
print(a)
```

```python
# we can use multiple for in List Comprehension
arr = [[1,2,3],[4,5,6],[9,4,5]]
p = [n for ele in arr for n in ele]
print(p)
c = [*arr[0],*arr[1],*arr[2]]
print(c)
```

```python
arr = [a+b for a in [2,4,5] for b in [5,6,7]]
print(arr)
b = [a*b for a in [3,4,5] for b in [6,7,5]]
print(b)
```

```python
arr = [[a+b for a in [2,4,5]] for b in [5,6,7]]
print(arr)
```
