```python
import numpy as np

arr = np.arange(2, 101)

primes = []

for num in arr:
    is_prime = True

    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            is_prime = False
            break

    if is_prime:
        primes.append(num)

print(np.array(primes))
```
